```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mean = pd.read_csv("noise_floor_mean.csv", header=0, index_col=0)
std = pd.read_csv("noise_floor_std.csv", header=0, index_col=0)
snr = mean/std
std.rename(columns={col: col+"std" for col in std.columns}, inplace=True)
snr.rename(columns={col: col+"snr" for col in snr.columns}, inplace=True)
df0 = pd.merge(mean, std, on='N')
df = pd.merge(df0, snr, on='N')
fig, ax = plt.subplots(1)
for col in snr.columns:
    df.plot(y=col, marker="o", logx=True, logy=True, ax=ax)
# df.to_csv("noisefloor.csv")
```
