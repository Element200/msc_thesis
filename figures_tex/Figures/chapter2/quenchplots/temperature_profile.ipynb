```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def T(x,t, T_0=10, Lambda=600, terms=100):
    s = 0
    for n in range(terms):
        s += np.sin((n+0.5)*np.pi*x)*np.exp(-Lambda*t*(2*n+1)**2)/(2*n+1)
    return 1+s*(T_0-1)*4/np.pi

x = np.linspace(0,1,1000)
plt.plot(T(x, 0))
out = []
for t in np.logspace(-5, -1, 20):
    plt.plot(T(x, t))
    out.append(T(x,t))
out = np.array(out)
```

```python
df = pd.DataFrame(out.T, columns=np.logspace(-5, -1, 20), index=x)
df.index.name='x'
```

```python
df.to_csv("temperature_profile.csv")
```
