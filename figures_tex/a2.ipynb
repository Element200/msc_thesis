```python
import mpemba

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
no_mpemba = mpemba.special_potentials.AsymmetricDoubleWellPotential(E_tilt=-0.2, force_asymmetry=1)
some_mpemba = mpemba.special_potentials.AsymmetricDoubleWellPotential(E_tilt=1, force_asymmetry=1)
strong_mpemba = mpemba.special_potentials.AsymmetricDoubleWellPotential(E_tilt=0.4, force_asymmetry=1)
```

```python
no_mpemba.plot_a_2s(10000)
some_mpemba.plot_a_2s(10000)
strong_mpemba.plot_a_2s(10000)
```

```python
k_BTs = np.logspace(np.log10(1), np.log10(10000), 500)
no_mpemba_a_2s = no_mpemba.a_k_boltzmannIC(k_BTs, n_x=1000)
some_mpemba_a_2s = some_mpemba.a_k_boltzmannIC(k_BTs, n_x=1000)
strong_mpemba_a_2s = strong_mpemba.a_k_boltzmannIC(k_BTs, n_x=1000)

df = pd.DataFrame(np.array([no_mpemba_a_2s, some_mpemba_a_2s, strong_mpemba_a_2s]).T, columns=['no', 'some', 'strong'], index=k_BTs)
df.index.name = 'T'
df
```

```python
df.to_csv('a2s.csv')
```
